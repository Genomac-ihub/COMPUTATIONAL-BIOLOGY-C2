```python
organism_name = "Arabidopsis thaliana"
gene_count = 27500
genome_size = 135000000  # base pairs
avg_gene_length = genome_size / gene_count
print(f"The genome of {organism_name} has approximately {gene_count} genes with an average length of {avg_gene_length:.2f} base pairs.")
```

```python
sequences = ["ATCG", "GGGTA", "TACGTT", "ATGATG", "CGTAA", "TTGCA"]
seq_lengths = []
for seq in sequences:
    print("Sequence:", seq)
    print("Length:", len(seq))
    print("G count:", seq.count('G'))
    seq_lengths.append(len(seq))
```

```python
def gc_content(seq):
    gc = seq.count("G") + seq.count("C")
    return (gc / len(seq)) * 100
for seq in sequences:
    gc = gc_content(seq)
    print(f"{seq}: GC Content = {gc:.2f}% ->", "High GC" if gc > 50 else "Low GC")
```

```python
gene_expression = {
    "BRCA1": 80.2,
    "TP53": 42.3,
    "EGFR": 95.7,
    "MYC": 23.0,
    "PTEN": 50.5
}
for gene, expr in gene_expression.items():
    if expr > 50:
        print(f"{gene} is highly expressed: {expr}")
```

```python
def reverse_complement(seq):
    complement = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    return ''.join([complement[base] for base in reversed(seq.upper())])
print(reverse_complement("ATGC"))
print(reverse_complement("GGATCC"))
print(reverse_complement("CGTACGTT"))
```

```python
gene_data = {
    "BRCA1": 80.2, "TP53": 42.3, "EGFR": 95.7, "MYC": 23.0, "PTEN": 50.5,
    "CDK2": 65.0, "RB1": 38.4, "AKT1": 88.9, "KRAS": 74.2, "APC": 59.1
}
sorted_genes = sorted(gene_data.items(), key=lambda x: x[1], reverse=True)
print("Gene ID | Expression | Status")
for gene, expr in sorted_genes:
    status = "Low" if expr < 40 else "Moderate" if expr <= 70 else "High"
    print(f"{gene} | {expr} | {status}")
```
