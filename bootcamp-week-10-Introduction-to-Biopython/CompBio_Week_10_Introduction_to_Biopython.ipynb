### Complement and Reverse Complement

Nucleotide sequence can be reverse complemented to get new sequence.
Also, the complemented sequence can be reverse complemented to get the original sequence. Biopython provides two methods to do this functionality − complement and reverse_complement.

```python
comp_seq = Seq('GATCGATGGGCCTATATAGGATCGAAAATCGC')

print(comp_seq + ' : ' + comp_seq.complement())   #To get the complement of the sequence

rc_seq = Seq('CTAGCTACCCGGATATATCCTAGCTTTTAGCG')
print(rc_seq +' : ' + rc_seq.reverse_complement())    # To get the reverse complement of the sequence

# Another way to just reverse a Seq object (or a Python string) is slice it with -1 step:
slice_seq = Seq('GCGATTTTCGATCCTATATAGGCCCATCGATC')
print(slice_seq + ' : ' + slice_seq[::-1])
```

### Transcription

Transcription is the process of changing DNA sequence into RNA sequence. The actual biological transcription process is performing a reverse complement (TCAG → CUGA) to get the mRNA considering the DNA as template strand. However, in bioinformatics and so in Biopython, we typically work directly with the coding strand and we can get the mRNA sequence by changing the letter T to U.\n



```python
from Bio.Seq import transcribe

dna_seq = Seq("ATGCCGATCGTAT")
transcribed_seq = transcribe(dna_seq)
print (f'coding sequence is : ', dna_seq )
print('Transcribed sequence is:' + transcribed_seq)
```

### Sequence annotation objects

The SeqRecord (Sequence Record) class is defined in the Bio.SeqRecord module. This class allows higher level features such as identifiers and features to be associated with a sequence, and is the basic data type for the Bio.SeqIO sequence input/output interface.

The SeqRecord class itself is quite simple, and offers the following information as attributes:

.seq
The sequence itself, typically a Seq object.

.id
The primary ID used to identify the sequence – a string. In most cases this is something like an accession number.

.name
A “common” name/id for the sequence – a string. In some cases this will be the same as the accession number, but it could also be a clone name. Think of this as being analogous to the LOCUS id in a GenBank record.

.description
A human readable description or expressive name for the sequence – a string.

.letter_annotations
Holds per-letter-annotations using a (restricted) dictionary of additional information about the letters in the sequence. The keys are the name of the information, and the information is contained in the value as a Python sequence (i.e. a list, tuple or string) with the same length as the sequence itself. This is often used for quality scores (e.g. Section Simple quality filtering for FASTQ files) or secondary structure information (e.g. from Stockholm/PFAM alignment files).

.annotations
A dictionary of additional information about the sequence. The keys are the name of the information, and the information is contained in the value. This allows the addition of more “unstructured” information to the sequence.

.features
A list of SeqFeature objects with more structured information about the features on a sequence (e.g. position of genes on a genome, or domains on a protein sequence).

.dbxrefs
A list of database cross-references as strings.

#### **BLAST Search**

Do not contact the server more often than once every 10 seconds.

Do not poll for any single RID more often than once a minute.

#### BLAST arguments

The qblast function has three non-optional arguments:

The first argument is the BLAST program to use for the search, as a lower case string. The programs and their options are described at the NCBI BLAST web page (https://blast.ncbi.nlm.nih.gov/Blast.cgi). Currently qblast only works with blastn, blastp, blastx, tblast and tblastx.

The second argument specifies the databases to search against. Again, the options for this are available on NCBI’s BLAST Help pages.

The third argument is a string containing your query sequence. This can either be the sequence itself, the sequence in fasta format, or an identifier like a GI number.

**Note** that the default settings on the NCBI BLAST website are not quite the same as the defaults on QBLAST.

```python
##   BLAST Tool	     Query Type	              Database Type	               Primary Use

##   blastn	         Nucleotide	              Nucleotide	                 Identifying homologous nucleotide sequences.
##   blastp	         Protein	                Protein	                     Finding similar proteins or functional conservation.
##   blastx	         Nucleotide	              Protein	                     Predicting proteins encoded by nucleotide sequences.
##   tblastn	       Protein	                Nucleotide (translated)	     Identifying nucleotide sequences encoding proteins similar to the query.
##   tblastx	       Nucleotide (translated)	Nucleotide (translated)	     Comparing highly divergent nucleotide sequences by translated protein similarity.
```

```python
from Bio.Blast import NCBIWWW
from Bio import SeqIO

# Read a sequence from a FASTA file
fasta_file = "/content/NM_001185075.fasta"
sequence = SeqIO.read(fasta_file, 'fasta')

# Perform BLAST Search
result_handle = NCBIWWW.qblast('blastn', 'nt', sequence.seq)

# Save result to a file
with open('blast_result.xml', 'w') as out_handle:
    out_handle.write(result_handle.read())
result_handle.close()
print('BLAST search complete.')
```

NCBIWWW: This module allows you to access NCBI’s BLAST service via the web (NCBI BLAST API). It is used for submitting BLAST queries programmatically.

Reading the Sequence from a FASTA File:

fasta_file: Specifies the path to the FASTA file containing the nucleotide sequence.
SeqIO.read():
Reads a single sequence record from the file in FASTA format.
The resulting sequence is a SeqRecord object that contains:
sequence.id: Identifier of the sequence (e.g., >NM_001185075).
sequence.seq: The actual nucleotide sequence.
sequence.description: Additional information in the header line of the FASTA file.
Purpose: This step loads the nucleotide sequence for the BLAST search.

Performing the BLAST Search


NCBIWWW.qblast():
Submits a BLAST query to the NCBI BLAST server.
Parameters:
'blastn': Specifies the type of BLAST search. Here, it is a nucleotide vs. nucleotide search (blastn).
'nt': Specifies the database to search against. Here, it is the nucleotide database (nt).
sequence.seq: The query sequence (nucleotide sequence extracted from the FASTA file).
result_handle:
Holds the raw result of the BLAST query in the form of a file-like object. This result is in XML format and can be parsed or saved for further analysis.

Saving the Result to a File:

open('blast_result.xml', 'w'):
Opens (or creates) a file named blast_result.xml in write mode.
This file will store the BLAST search results.
result_handle.read():
Reads the content of the result (in XML format) from the result_handle object.
out_handle.write():
Writes the content (BLAST search results) into the file.
result_handle.close():
Closes the result_handle to release resources.

Printing Completion Message:

Simply prints a message to the console to indicate that the BLAST search is complete and the results are saved in the specified file.
Key Notes
Input Data:

The FASTA file should contain a nucleotide sequence you want to query against the NCBI nucleotide database (nt).
Output Data:

The BLAST search result is saved as an XML file (blast_result.xml), which can be parsed or visualized using Biopython or other tools.
Why Use XML Format?:

XML is the standard output format for BLAST results. It can be easily parsed by Biopython’s NCBIXML module for further analysis (e.g., retrieving alignments, scores, or e-values).
BLAST Search Type:

This script performs a nucleotide-to-nucleotide (BLASTN) search, making it ideal for querying DNA or RNA sequences against databases of nucleotide sequences.


#### Opening the xml file using code

```python
from Bio.Blast import NCBIXML

# Parse the BLAST result XML file
result_handle = open("blast_result.xml")
blast_record = NCBIXML.read(result_handle)
result_handle.close()

# Print some information about the BLAST hits
print("No of alignments:", len(blast_record.alignments))

# Iterate over each alignment and print some details
for alignment in blast_record.alignments[:5]:   # Print details for the first five alignments
    print("Alignment title:", alignment.title)
    print("Number of high-scoring segment pairs (hsps):", len(alignment.hsps))
    for hsp in alignment.hsps[:1]:  # Print details for the first HSP of each alignment
        print("Query start:", hsp.query_start)
        print("Query end:", hsp.query_end)
        print("Hit start:", hsp.sbjct_start)
        print("Hit end:", hsp.sbjct_end)
        print("Alignment length:", hsp.align_length)
        print("E-value:", hsp.expect)
        print("Bit score:", hsp.bits)
        print("Query sequence:", hsp.query)
        print("Hit sequence:", hsp.sbjct)
```
