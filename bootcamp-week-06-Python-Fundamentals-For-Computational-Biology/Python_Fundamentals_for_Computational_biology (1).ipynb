## **Slicing**

We use slicing when we wish to return a sequence of values from a list instead of just a single value.

Syntax: list[start:end]

start: The index from which the slicing begins (included).

end: The index at which the slicing ends (not included).

step: The amount of steps to take for each slice.

For instance, if we want to slice from the 3rd to the 5th element, the start index will be 2 (to be included), and the end index will be 5 (not to be included).

[start:stop:step] = [9]

[step:] = [::9]

[start:stop] = [:9]

start is the index you want to start calling/slicing from. stop is the index you want to stop the calling/slicing. step is the amount of jumps/steps you want the slicing to take per slice.

# Conditional statements in Python

Conditional statements in programming are used to execute specific blocks of code based on whether certain conditions are met.

**if, elif, and else**

**if** statements are the most basic form of conditional execution. They are used to execute a block of code only if a specified condition is True.

```python
# Check how many digits our variable has
x = 21

if x > 20:
    print("Our variable is a two-digit number")
```

```python
#check how many digits our variable has
y = 3000

if y > 9990:
    print("Our variable is of 4-digits")
```

```python
# Let's try it with booleans
x = 0

if x:  # "if x" is the same as "if x==True"
    print("x was true")
```

```python
x = False

if x:
    print("x was true") # It did not print becuase x is False
```

**Nested if** statements are used when we need to perform another conditional check within an if block.

```python
x = 15
odd_number = True

if x > 9:
    if odd_number:  # remember that "if odd_number" is equivalent to "if odd_number == True"
        print("x is an odd number and a two-digit number")
```

```python
z = 13
even_number = False

if z > 10:
    if even_number == False:
        print("z is an odd number and a two-digit number")
```

**if-else statements** execute one block of code if the condition is True and another block if it is False. We can use multiple "if" but only one "else". "else" cannot be used without an "if".

```python
x = 15

if x < 10:
    print('10')
else:
    print('Not 10')
```

```python
x = 5

if x < 20:
    print("x is a middle-number")
else:
    print("x is not a midle-number")
```

```python
x = True

if x:
    print("x was true")
else:
    print("y will be printed in any case where x is not true")
```

```python
# GC content for multiple sequences

sequences = ["ATGCGT", "TTTAAA", "GGCCGG"]

for seq in sequences:
    gc = (seq.count("G") + seq.count("C")) / len(seq) * 100
    print(f"GC content of {seq}: {gc:.2f}%")
```

```python
# Check if any sequence contains a start codon

for seq in sequences:
    if "ATG" in seq:
        print(f"Start codon found in {seq}")
```

```python
# Transcribe multiple DNA sequences to RNA

for seq in sequences:
    rna = seq.replace("T", "U")
    print(f"RNA: {rna}")
```

```python
# Flagging sequences with low GC content

for seq in sequences:
    gc = (seq.count("G") + seq.count("C")) / len(seq) * 100
    if gc < 40:
        print(f"{seq}: Low GC")
    else:
        print(f"{seq}: OK GC")
```

**Loop control statements** are used to change the execution flow of loops. The primary loop control statements in Python are **break, continue, and pass.**

**break** immediately stops the execution of a loop, and the control flows to the statement immediately after the loop.

**continue** skips the rest of the loop's body for the current iteration and moves to the next iteration.

**pass** does nothing. It's used as a placeholder.

```python
x = 1
while x < 10:
    print('x is currently: ', x)
    print('x is still less than 10, adding 1 to x')
    x+=1
    if x==4:
        print('breaking because x==4')
        break
```

**User-defined functions** are functions that we create to perform specialised tasks based on our unique requirements.

We can break down the creation of a function into five components;

1.) Function **definition**: we use the keyword **"def"** to define a function, followed by round brackets and a colon. The colon is used to indicate the start of a new block of code.

2.) Function **name**: we need to give our function a name which we will use to call it later. There are some guidelines on how to name a function, such as the name must begin with a letter or an underscore, it must be in lowercase, it can have numbers and be any length (it's best to keep it short), it cannot be the same as any Python keyword, and must describe the function's purpose.

3.) Function **inputs**: Inputs, also known as parameters, define the variables required for the function to perform its tasks. It's worth noting that we don't have to submit an input every time. If the function tasks do not require any inputs, we can leave this field empty. It is essential to differentiate between function parameters and arguments.

4.) Function **instructions/tasks**: is the block of code within the function that serves as the set of instructions that the function must follow. To indicate that the code that follows belongs to the function, we type it one tab away from the declaration line.

5.) **Return statement**: defines what the result of the function or the output should be.

Note: A return statement is not required for the function to be created successfully. However, because every function in Python must return something, functions that do not have a return statement return None by default.

```python
#Create the calc_forest_area function with docstrings
def calc_forest_area(length, width):
    """
    Calculates the area of a given forest by multiplying its length and width

    Parameters
    -------
    length: The length of the forest in unit
    width: The width of the forest in unit

    Return
    -------
    An integer that represents the area of the forest in unit squared

    """
    area = length * width
    return area
```

```python
def nucleotide_segmentor(input1):
    """
    Identifies whether each nucleotide in the input sequence is a purine or pyrimidine.

    Parameters:
    ----------
    input1 : str
        A DNA sequence consisting of nucleotide characters (A, T, G, C).

    Returns:
    --------
    None
        Prints whether each nucleotide is a purine or pyrimidine.
    """
    for base in input1:
        if base == "A" or base == "G":
            print(f"{base} is a purine")
        elif base == "T" or base == "C":
            print(f"{base} is a pyrimidine")
        else:
            print(f"{base} is not a valid DNA nucleotide")
```

```python
# Motif Finding with Loops and Functions

def find_motif(sequence, motif):
    positions = []
    for i in range(len(sequence) - len(motif) + 1):
        if sequence[i:i+len(motif)] == motif:
            positions.append(i)
    return positions
```

## **Basic File Handling**

How to read sequences from a .txt file.

```python
# Example

with open("dna_sequences.txt") as f:
    sequences = f.readlines()
```

**with open("dna_sequences.txt") as f:**
This is a context manager for safely opening a file.

**"dna_sequences.txt"** is the file being opened (must be in the same directory unless a full path is provided).

**as f** means the file object is temporarily named f.

**Automatic cleanup:** When the block finishes, Python will automatically close the file, even if there’s an error inside the block.

The **with statement** is used to manage resources like files, network connections, or database sessions — things that need to be opened and then properly closed after use.


You're telling Python:

"Open this file, let me use it inside this block, and when I’m done — even if there’s an error — make sure to close it automatically."

without with:
You'd have to do it manually:

```python
f = open("dna_sequences.txt")
sequences = f.readlines()
f.close()  # You must remember to do this
```

If you forget **f.close()**, it can lead to:

Memory leaks

Locked files

Too many open files error in larger *programs*

**f.readlines()**
Reads all the lines in the file into a list.

Each line becomes a string in the list, including the newline character \n at the end.

**Optional: Remove newline characters**

To clean the sequences:

```python
sequences = [line.strip() for line in sequences]
```
